### 1. Xác suất lý thuyết

Các kết quả lớn hơn 18 là các số từ 19 đến 36, có 18 kết quả.

Do đó, xác suất lý thuyết là:

P(X>18)= 18/37 = 48.65%.

```python
import numpy as np

rng = np.random.default_rng(2026)

for n in [100, 1000, 10000]:
    spins = rng.integers(0, 37, size=n)
    is_greater_18 = spins > 18
    proportion = np.mean(is_greater_18)

    print(f"So luot quay: {n}")
    print(f"Ty le thuc nghiem: {proportion:.4f}")
    print(f"Ty le phan tram: {proportion:.2%}")
    print()

print(f"Xac suat ly thuyet: {18/37:.4f}")
```

### 2. Giải thích chương trình

- `np.random.default_rng(2026)`: Khởi tạo bộ sinh số ngẫu nhiên với seed 2026 để có thể tái lập kết quả.
- `rng.integers(0, 37, size=n)`: Mô phỏng `n` lượt quay, mỗi lượt nhận một số nguyên từ 0 đến 36.
- `spins > 18`: Tạo một mảng giá trị logic. Giá trị `True` nếu kết quả lớn hơn 18 và `False` nếu ngược lại.
- `np.mean(is_greater_18)`: Tính tỷ lệ kết quả lớn hơn 18 trong mẫu mô phỏng.
- `18 / 37`: Tính xác suất lý thuyết.

### 3. Nhận xét

Xác suất lý thuyết để kết quả roulette lớn hơn 18 là 18/37 = 48.65%

Tỷ lệ thực nghiệm có thể khác xác suất lý thuyết do tính ngẫu nhiên của quá trình mô phỏng. Khi số lượt quay tăng, tỷ lệ thực nghiệm thường có xu hướng tiến gần xác suất lý thuyết theo luật số lớn. Tuy nhiên, tỷ lệ không nhất thiết tiến gần hơn ở mọi lần tăng kích thước mẫu.
